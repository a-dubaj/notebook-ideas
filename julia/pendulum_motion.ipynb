```julia
using Plots
using DifferentialEquations

function pendulum!(du, u, p, t)
    θ, ω = u
    g, L = p
    du[1] = ω
    du[2] = -(g / L) * sin(θ)
end

g = 9.81
L = 1.0
p = (g, L)

u0 = [π / 4, 0.0]
tspan = (0.0, 10.0)
prob = ODEProblem(pendulum!, u0, tspan, p)
sol = solve(prob, Tsit5(), saveat=0.01)

θ_vals = [u[1] for u in sol.u]
t_vals = sol.t

plot(t_vals, θ_vals,
    xlabel="Time (s)",
    ylabel="Angle θ (rad)",
    title="Simple Pendulum Motion",
    label="θ(t)",
    linewidth=2)
```
