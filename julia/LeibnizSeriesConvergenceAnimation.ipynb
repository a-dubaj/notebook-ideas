```julia
using Plots
using Printf

gr()  # fast backend, good for animations

# Partial sum of the Leibniz series: s_n = sum_{k=0}^{n-1} (-1)^k / (2k+1)
function leibniz_partial_sums(N)
    s = zeros(Float64, N)
    acc = 0.0
    for k in 0:N-1
        acc += (-1)^k / (2k + 1)
        s[k+1] = 4 * acc
    end
    return s
end

N = 300                      # number of terms to animate (more = slower render)
partial_sums = leibniz_partial_sums(N)
errors = abs.(partial_sums .- Float64(pi))

anim = @animate for n in 1:N
    p1 = plot(1:n, partial_sums[1:n],
        xlim = (0, N), ylim = (2.6, 3.6),
        xlabel = "Number of terms n", ylabel = "Partial sum",
        label = "s_n (Leibniz)", lw = 2, color = :steelblue,
        title = @sprintf("s_%d = %.6f   (π = %.6f)", n, partial_sums[n], pi))
    hline!(p1, [pi], label = "π", color = :crimson, lw = 1.5, ls = :dash)
    scatter!(p1, [n], [partial_sums[n]], color = :orange, ms = 5, label = "")

    p2 = plot(1:n, errors[1:n],
        xscale = :log10, yscale = :log10,
        xlim = (1, N), ylim = (minimum(errors) * 0.5, 2),
        xlabel = "n", ylabel = "Absolute error",
        label = "error", lw = 2, color = :seagreen,
        marker = :circle, ms = 3)

    plot(p1, p2, layout = (1, 2), size = (900, 400))
end

# Display inline in the notebook (no filename -> saves to a temp file and shows the GIF player)
gif(anim, fps = 20)
```
