```julia
using Plots

setprecision(BigFloat, 100; base=10)

# True value of π at high precision, used as reference
true_pi = BigFloat(π)

function leibniz_pi(n)
    setprecision(BigFloat, 100; base=10)
    s = BigFloat(0)
    for k in 0:n-1
        s += (-1)^k / BigFloat(2k + 1)
    end
    return 4 * s
end

n_terms = [10^i for i in 1:6]
errors = [abs(leibniz_pi(n) - true_pi) for n in n_terms]

plot(n_terms, Float64.(errors),
    xscale=:log10, yscale=:log10,
    xlabel="Number of series terms",
    ylabel="Absolute error",
    title="Convergence of the Leibniz series to π",
    marker=:circle,
    label="error |π_approx - π|",
    legend=:topright)
```
