```fortran
program pi_convergence
    implicit none
    integer :: n_terms
    real(8) :: pi_approx, term
    integer :: unit_num

    open(newunit=unit_num, file='pi_convergence.dat', status='replace')

    pi_approx = 0.0_8
    do n_terms = 1, 100000
        term = ((-1.0_8)**(n_terms - 1)) / (2.0_8 * n_terms - 1.0_8)
        pi_approx = pi_approx + term
        if (mod(n_terms, 500) == 0) then
            write(unit_num, '(I8, 1X, F20.15)') n_terms, 4.0_8 * pi_approx
        end if
    end do

    close(unit_num)

    open(newunit=unit_num, file='plot_script.gp', status='replace')
    write(unit_num, '(A)') "set terminal png size 800,500"
    write(unit_num, '(A)') "set output 'pi_convergence.png'"
    write(unit_num, '(A)') "set xlabel 'Number of terms'"
    write(unit_num, '(A)') "set ylabel 'Approximation of pi'"
    write(unit_num, '(A)') "set title 'Leibniz series convergence to pi'"
    write(unit_num, '(A)') "plot 'pi_convergence.dat' using 1:2 " // &
        "with lines title 'Approximation', " // &
        "3.14159265358979 with lines title 'pi (reference)'"
    close(unit_num)

    call execute_command_line("gnuplot plot_script.gp")

    print *, "Plot saved to pi_convergence.png"
end program pi_convergence
```
