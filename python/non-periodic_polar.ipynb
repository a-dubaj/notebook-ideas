```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

DT = 0.1
T1 = np.arange(0, 333, DT)
A_RAD = np.sin(np.pi * T1) + np.sin(3.5 * T1) 

def f(x):
    return np.sin(np.pi * x) + np.sin(3.5 * x)

fig, ax = plt.subplots(figsize=(8, 8), dpi=120)
fig.patch.set_facecolor("black")
ax.set_facecolor("black")
ax.set_xlim(30, 470)
ax.set_ylim(30, 470)
ax.set_aspect("equal")
ax.axis("off")

line, = ax.plot([], [], 'wo', markersize=1.5)

def update(frame):
    offset = 1.0 + frame * 0.002
    t2 = offset + np.arange(len(T1)) * DT

    r = f(t2)

    x = 250 + 110 * r * np.cos(A_RAD)
    y = 250 + 110 * r * np.sin(A_RAD)

    line.set_data(x, y)
    return line,

ani = FuncAnimation(
    fig, 
    update, 
    frames=1000, 
    interval=20, 
    blit=True,
    repeat=True
)

plt.show()
```
