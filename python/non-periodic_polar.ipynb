```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

dt = 0.1
offset = 1.0

t1 = np.arange(0, 333, dt)

def f(x):
    return np.sin(np.pi*x) + np.sin(3.5*x)

fig, ax = plt.subplots(figsize=(5,5))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")
ax.set_xlim(30,470)
ax.set_ylim(30,470)
ax.set_aspect("equal")
ax.axis("off")

scatter = ax.scatter([], [], s=0.5, c="white")

def update(frame):
    global offset

    t2 = offset + np.arange(len(t1))*dt

    r = f(t2)
    a = f(t1)

    x = 250 + 110*r*np.cos(a)
    y = 250 + 110*r*np.sin(a)

    scatter.set_offsets(np.c_[x, y])

    offset += 0.002
    return scatter,

ani = FuncAnimation(fig, update, interval=20, blit=True)

plt.show()
```
