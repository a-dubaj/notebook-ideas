```python
import matplotlib.pyplot as plt
import numpy as np

def leibniz_pi(n_terms):
    approx = 0
    approximations = []
    for k in range(n_terms):
        approx += ((-1)**k) / (2*k + 1)
        approximations.append(4 * approx)
    return approximations

n = 1000
approx = leibniz_pi(n)

plt.figure(figsize=(8, 5))
plt.plot(approx, label='Leibniz approximation')
plt.axhline(y=np.pi, color='red', linestyle='--', label='pi (actual value)')
plt.xlabel('Number of terms')
plt.ylabel('Value')
plt.title('Convergence of the Leibniz series to pi')
plt.legend()
plt.show()
```
