```python
import numpy as np
import matplotlib.pyplot as plt

n_points = 5000
x = np.random.uniform(-1, 1, n_points)
y = np.random.uniform(-1, 1, n_points)
inside = x**2 + y**2 <= 1

pi_estimate = 4 * np.sum(inside) / n_points

plt.figure(figsize=(6, 6))
plt.scatter(x[inside], y[inside], color='steelblue', s=2, label='Inside circle')
plt.scatter(x[~inside], y[~inside], color='lightgray', s=2, label='Outside circle')
plt.title(f'Monte Carlo estimation: pi ≈ {pi_estimate:.5f}')
plt.legend()
plt.axis('equal')
plt.show()
```
